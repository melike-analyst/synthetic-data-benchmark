```python
import pandas as pd
import time
from sdv.metadata import SingleTableMetadata
from sdv.single_table import GaussianCopulaSynthesizer
```

```python

def run_gaussian_copula(train_path, output_path, dataset_name):
    train_df = pd.read_csv(train_path)

    # SDV'nin veri şemasını otomatik algılamasını sağlıyoruz
    metadata = SingleTableMetadata()
    metadata.detect_from_dataframe(train_df)

    start = time.time()
    synthesizer = GaussianCopulaSynthesizer(metadata)
    synthesizer.fit(train_df)
    synthetic_data = synthesizer.sample(num_rows=len(train_df))
    elapsed = time.time() - start

    synthetic_data.to_csv(output_path, index=False)
    print(f"[{dataset_name}] Gaussian Copula tamamlandı. Süre: {elapsed:.2f} sn")
    return elapsed

t1 = run_gaussian_copula(
    '../data/raw/adult_train.csv',
    '../data/synthetic/adult_gaussian_copula.csv',
    'Adult'
)

t2 = run_gaussian_copula(
    '../data/raw/creditcard_train.csv',
    '../data/synthetic/creditcard_gaussian_copula.csv',
    'CreditCard'
)
```
