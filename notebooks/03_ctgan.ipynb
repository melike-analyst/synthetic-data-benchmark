```python
import pandas as pd
import time
from sdv.metadata import SingleTableMetadata
from sdv.single_table import CTGANSynthesizer
```

```python
def run_ctgan(train_path, output_path, dataset_name, epochs=300):
    train_df = pd.read_csv(train_path)
    metadata = SingleTableMetadata()
    metadata.detect_from_dataframe(train_df)

    start = time.time()
    synthesizer = CTGANSynthesizer(metadata, epochs=epochs, verbose=True)
    synthesizer.fit(train_df)
    synthetic_data = synthesizer.sample(num_rows=len(train_df))
    elapsed = time.time() - start

    synthetic_data.to_csv(output_path, index=False)
    print(f"[{dataset_name}] CTGAN tamamlandı. Süre: {elapsed:.2f} sn")
    return elapsed

t3 = run_ctgan(
    '../data/raw/adult_train.csv',
    '../data/synthetic/adult_ctgan.csv',
    'Adult'
)

t4 = run_ctgan(
    '../data/raw/creditcard_train.csv',
    '../data/synthetic/creditcard_ctgan.csv',
    'CreditCard',
    epochs=150  # CreditCard 284K satır, GPU'suz makinede epoch sayısını düşür
)
```
