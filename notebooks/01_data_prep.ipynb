```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# --- Adult Income ---
df_adult = pd.read_csv('../data/raw/adult.csv')
print(df_adult.shape)
print(df_adult.dtypes)
print(df_adult['income'].value_counts(normalize=True))

# Temizlik: '?' değerlerini NaN yap, at
df_adult = df_adult.replace('?', pd.NA).dropna()

# Holdout ayrımı — %70 "gerçek kaynak" (sentetik üretimde kullanılacak),
# %30 hiçbir üretim sürecine karışmayan "temiz" test seti
train_adult, holdout_adult = train_test_split(
    df_adult, test_size=0.3, random_state=42, stratify=df_adult['income']
)

train_adult.to_csv('../data/raw/adult_train.csv', index=False)
holdout_adult.to_csv('../data/holdout/adult_holdout.csv', index=False)
print(f"Train: {train_adult.shape}, Holdout: {holdout_adult.shape}")
```

```python
# --- Credit Card Fraud ---
df_cc = pd.read_csv('../data/raw/creditcard.csv')
print(df_cc.shape)
print(df_cc['Class'].value_counts(normalize=True))

train_cc, holdout_cc = train_test_split(
    df_cc, test_size=0.3, random_state=42, stratify=df_cc['Class']
)

train_cc.to_csv('../data/raw/creditcard_train.csv', index=False)
holdout_cc.to_csv('../data/holdout/creditcard_holdout.csv', index=False)
print(f"Train: {train_cc.shape}, Holdout: {holdout_cc.shape}")
```
