```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
results_df = pd.read_csv('../results/raw_results.csv')

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.barplot(data=results_df, x='method', y='fidelity_score', hue='dataset', ax=axes[0])
axes[0].set_title('Fidelity Score (yüksek = iyi)')

sns.barplot(data=results_df, x='method', y='utility_gap', hue='dataset', ax=axes[1])
axes[1].set_title('Utility Gap (düşük = iyi)')

plt.tight_layout()
plt.savefig('../results/figures/comparison_overview.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(7,6))
sns.scatterplot(data=results_df, x='privacy_dcr_mean', y='fidelity_score',
                 hue='method', style='dataset', s=150)
plt.title('Fidelity vs. Privacy Trade-off')
plt.savefig('../results/figures/tradeoff.png', dpi=150)
plt.show()
```
